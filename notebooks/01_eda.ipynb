# 01 - Analyse exploratoire (EDA) de l'or

Ce notebook appelle les mêmes fonctions que `gold-agents eda` et affiche les figures. Les seuils et tests sont calculés sur le **train seulement**.

Prérequis : `uv sync --extra dev --extra eda` puis `uv run --with jupyter jupyter lab`.

```python
import os
from pathlib import Path

from IPython.display import Image, Markdown, display

from gold_agents.config import load_config
from gold_agents.eda.report import run_eda, summary_lines

# Les chemins de config.yaml sont relatifs à la racine du dépôt.
if Path.cwd().name == "notebooks":
    os.chdir(Path.cwd().parent)

cfg = load_config()
result = run_eda(cfg, Path("outputs/eda"))
print("\n".join(summary_lines(result.summary)))
```

```python
for path in result.figure_paths:
    display(Markdown(f"### {path.stem}"), Image(filename=str(path)))
```
